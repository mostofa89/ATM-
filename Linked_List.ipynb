```python
def plusMinus(arr):
    # Write your code here
    zero = 0
    positive = 0
    negative = 0
    for num in arr:
        if num == 0:
            zero += 1
        elif num > 0:
            positive += 1

        else:
            negative += 1
    output = f"""{(positive / len(arr)):.6f}
{(negative / len(arr)):.6f}
{(zero / len(arr)):.6f}"""

    return output


print(plusMinus([1, -1, 0, 1, -1]))
```
